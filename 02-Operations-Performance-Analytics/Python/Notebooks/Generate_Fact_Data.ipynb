```python
dim_employee = pd.read_sql("SELECT * FROM dim_employee", engine)
dim_date = pd.read_sql("SELECT * FROM dim_date", engine)
dim_team = pd.read_sql("SELECT * FROM dim_team", engine)
dim_manager = pd.read_sql("SELECT * FROM dim_manager", engine)
dim_process = pd.read_sql("SELECT * FROM dim_process", engine)
dim_country = pd.read_sql("SELECT * FROM dim_country", engine)
dim_shift = pd.read_sql("SELECT * FROM dim_shift", engine)
dim_leave = pd.read_sql("SELECT * FROM dim_leave", engine)

print("Employees :", len(dim_employee))
print("Dates     :", len(dim_date))
print("Teams     :", len(dim_team))
print("Managers  :", len(dim_manager))
print("Processes :", len(dim_process))
print("Countries :", len(dim_country))
print("Shifts    :", len(dim_shift))
print("Leaves    :", len(dim_leave))
```

```python
import random
import numpy as np
from faker import Faker

fake = Faker()

# Number of records
NUM_RECORDS = 100000

print("Records to generate:", NUM_RECORDS)
```

```python
employee_ids = dim_employee["employee_id"].tolist()
date_ids = dim_date["date_id"].tolist()
team_ids = dim_team["team_id"].tolist()
manager_ids = dim_manager["manager_id"].tolist()
process_ids = dim_process["process_id"].tolist()
country_ids = dim_country["country_id"].tolist()
shift_ids = dim_shift["shift_id"].tolist()
leave_ids = dim_leave["leave_id"].tolist()

print("Lookup lists ready!")
```

```python
fact_data = []

attendance_options = ["Present", "WFH", "Leave"]

for i in range(NUM_RECORDS):

    employee = random.choice(employee_ids)
    date = random.choice(date_ids)

    team = random.choice(team_ids)
    manager = random.choice(manager_ids)
    process = random.choice(process_ids)
    country = random.choice(country_ids)
    shift = random.choice(shift_ids)

    attendance = random.choices(
        attendance_options,
        weights=[85, 10, 5],
        k=1
    )[0]

    if attendance == "Leave":
        leave = random.choice(leave_ids)
        login_hours = 0
        productive_hours = 0
        idle_hours = 0
        overtime_hours = 0
        cases_completed = 0
        cases_reopened = 0
        aht = 0
        quality = 0
        csat = 0
        adherence = 0
        escalation = 0

    else:

        leave = None

        login_hours = round(random.uniform(7.5,10.5),2)

        productive_hours = round(
            login_hours * random.uniform(0.70,0.90),2
        )

        idle_hours = round(
            login_hours - productive_hours,
            2
        )

        overtime_hours = round(
            max(0, login_hours-9),
            2
        )

        cases_completed = random.randint(35,110)

        cases_reopened = random.randint(0,5)

        aht = round(random.uniform(4,12),2)

        quality = round(random.uniform(85,100),2)

        csat = round(random.uniform(3.5,5),2)

        adherence = round(random.uniform(85,100),2)

        escalation = random.randint(0,3)

    fact_data.append([

        date,
        employee,
        team,
        manager,
        process,
        country,
        shift,
        leave,

        login_hours,
        productive_hours,
        idle_hours,
        overtime_hours,

        cases_completed,
        cases_reopened,
        aht,
        quality,
        csat,
        adherence,
        attendance,
        escalation

    ])

print("Fact rows generated:", len(fact_data))
```

```python
fact_df = pd.DataFrame(
    fact_data,
    columns=[
        "date_id",
        "employee_id",
        "team_id",
        "manager_id",
        "process_id",
        "country_id",
        "shift_id",
        "leave_id",
        "login_hours",
        "productive_hours",
        "idle_hours",
        "overtime_hours",
        "cases_completed",
        "cases_reopened",
        "average_handle_time",
        "quality_score",
        "csat_score",
        "adherence_percent",
        "attendance_status",
        "escalation_count"
    ]
)

fact_df.head()
```
